## What is Bash?

- **Bash** (Bourne Again SHell) is a command-line shell used for interacting with the Unix/Linux operating system.

---

## Why Learn Bash Scripting?

- **Efficiency**: Combine and execute multiple commands in a single script.
- **Customizability**: Customize your workflow and environment.
- **Portability**: Write scripts that run across multiple Unix-like systems.

## Writing Your First Bash Script



### Steps to create a script

1. Write plain text in a text editor.
2. Use a portable Bash shebang: `#!/usr/bin/env bash`.
3. Save the file with a descriptive name, commonly ending in `.sh`.
4. Add owner execute permission with `chmod u+x script.sh`.
5. Run it as `./script.sh` and inspect its exit status and output.


### Example script

```bash
#!/usr/bin/env bash
# This is a simple Bash script.
printf 'Hello, World!\n'
```


Try it - now!

- **Explanation**:
  - `#!/usr/bin/env bash`: asks the environment to locate Bash.
  - `printf`: writes predictable formatted text.
  - Save this as `hello.sh`, run `chmod u+x hello.sh`, then execute `./hello.sh`.

# Introduction to Variables in Bash Scripting

## What is a Variable?

- A **variable** is a way to store data that can be referenced and manipulated within a script.
- Bash variables do not require explicit declaration of data types (e.g., string, integer). Everything is treated as a string.

---

---

## Basic Syntax

```bash
variable_name=value
```

- No spaces around the equal sign.
- **variable_name**: The name of the variable (should not start with a number).
- **value**: The data that is stored in the variable.

---

## Simple Example

```bash
greeting="Hello, World!"
echo $greeting
```

- **Output:**
  ```
  Hello, World!
  ```

---

---

## Accessing Variables

- Use the `$` symbol to access the value of a variable.

```bash

- **Output:**

  ```

## Variable types

### Shell variables

Variables exist in the current shell:

```bash
greeting="Hello"
printf '%s\n' "$greeting"
```

### Environment variables

`export` includes a variable in the environment inherited by child processes. It does not make the value system-wide:

```bash
export MY_VAR="Exported value"
printf '%s\n' "$MY_VAR"
```

### Positional parameters

`$1`, `$2`, and later values contain script or function arguments.

## Variable scope

Bash assignments inside a function are global to the current shell unless the variable is declared with `local`. This means a function can accidentally overwrite a variable created outside the function.

Compare these two functions:

```bash
message="original value"

change_global() {
    message="changed by the function"
}

change_local() {
    local message="local to this function"
    printf 'Inside change_local: %s\n' "$message"
}

printf 'Before change_global: %s\n' "$message"
change_global
printf 'After change_global: %s\n' "$message"

change_local
printf 'After change_local: %s\n' "$message"
```

Output:

```text
Before change_global: original value
After change_global: changed by the function
Inside change_local: local to this function
After change_local: changed by the function
```

`change_global` changes the outer `message` variable. `change_local` creates a separate `message` variable that only exists while the function is running.

# Introduction to For Loops in Bash Scripting

---

## What is a For Loop?

- A **for loop** is a control flow statement for specifying iteration.
- It allows you to execute a sequence of commands multiple times.
- Useful for automating repetitive tasks in scripts.

---



## Basic Syntax

```bash
for variable in list
do
    commands
done
```

- **variable**: A placeholder for each item in the list.
- **list**: A set of values (numbers, strings, etc.).

---



## Simple Example

```bash
for i in 1 2 3 4 5
do
    echo "Number: $i"
done
```

- **Output:**
  ```
  Number: 1
  Number: 2
  Number: 3
  Number: 4
  Number: 5
  ```

---



## Using Brace Expansion

- Simplify sequences with brace expansion `{start..end}`.

```bash
for i in {1..5}
do
    echo "Number: $i"
done
```

- Equivalent to listing numbers manually.

---



## Looping Over Files

- Iterate over files matching a pattern.

```bash
for file in *.txt
do
    echo "Processing $file"
done
```

- **Processes all `.txt` files** in the current directory.

---



## C-Style For Loops

- Similar to loops in languages like C or Java.

```bash
for (( i=1; i<=5; i++ ))
do
    echo "Number: $i"
done
```

- **Initialization**: `i=1`
- **Condition**: `i<=5`
- **Increment**: `i++`

---



## Nested For Loops

- Loops within loops for multi-level iteration.

```bash
for i in {1..3}
do
    for j in a b
    do
        echo "$i$j"
    done
done
```

- **Output:**
  ```
  1a
  1b
  2a
  2b
  3a
  3b
  ```

---



## Practical Example: Renaming Files

- Append a prefix to multiple files.

```bash
for file in *.jpg
do
    mv "$file" "new_$file"
done
```

- **Renames**: `image.jpg` to `new_image.jpg`

---



## Reading a file one line at a time

Do not use `for user in $(cat users.txt)` for general records. Command substitution performs word splitting and glob expansion.

```bash
while IFS= read -r user; do
    printf 'Processing user: %s\n' "$user"
done < users.txt
```

This keeps spaces and backslashes in each input line intact.


## Loop practices

- Quote expansions: `"$file"`.
- Prefer descriptive variable names.
- Preview changes with `printf` before using `mv` or `rm`.
- Handle the case where a glob matches no files:

```bash
for file in ./*.txt; do
    [[ -e "$file" ]] || continue
    printf 'Processing %s\n' "$file"
done
```


---

## Conclusion

- For loops are essential for automation in Bash scripting.
- Multiple syntaxes to fit different scenarios.
- Enhance scripts by combining loops with other control structures.

# Introduction to If Statements in Bash Scripting

---

## What is an If Statement?

- An **if statement** allows your script to make decisions based on conditions.
- Executes commands **only if** a specified condition is true.
- Essential for adding logic and control flow to your scripts.

---

## Basic Syntax

```bash
if [ condition ]
then
    commands
fi
```

- **condition**: A test expression that returns true or false.
- **commands**: One or more commands to execute if the condition is true.

---



## Simple Example

```bash
if [ -f "file.txt" ]
then
    echo "File exists."
fi
```

- **Explanation**:
  - `-f "file.txt"` checks if `file.txt` is a regular file.
- **Output (if file exists):**
  ```
  File exists.
  ```

---



## Using Else

- Execute alternative commands if the condition is false.

```bash
if [ condition ]
then
    commands_if_true
else
    commands_if_false
fi
```

---

### Example with Else

```bash
if [ -d "/path/to/directory" ]
then
    echo "Directory exists."
else
    echo "Directory does not exist."
fi
```

- **Explanation**:
  - Checks if the specified directory exists.
  - Outputs a message based on the result.

---



## Using Elif (Else If)

- Check multiple conditions in sequence.

```bash
if [ condition1 ]
then
    commands1
elif [ condition2 ]
then
    commands2
else
    commands3
fi
```

---

### Example with Elif

```bash
if [ $score -ge 90 ]
then
    echo "Grade: A"
elif [ $score -ge 80 ]
then
    echo "Grade: B"
else
    echo "Grade: C"
fi
```

- **Explanation**:
  - Assigns grades based on the value of `score`.

---



## Comparison Operators

### Numeric Comparison

- `-eq`: equal to
- `-ne`: not equal to
- `-gt`: greater than
- `-lt`: less than
- `-ge`: greater than or equal to
- `-le`: less than or equal to

### String Comparison

- `=`: strings are equal
- `!=`: strings are not equal
- `-z string`: string is null (empty)
- `-n string`: string is not null

---


## Logical operators

With `[[ ... ]]`, use `&&`, `||`, and `!`:

```bash
if [[ $age -ge 18 && $age -lt 65 ]]; then
    echo "Eligible for work."
fi
```

Avoid the older `-a` and `-o` test operators in new examples because their parsing is harder to reason about.



## Using Double Brackets `[[ ]]`

- Offers advanced features like pattern matching.

```bash
if [[ $name == "Alice" ]]
then
    echo "Hello, Alice!"
fi
```

- **Advantages**:
  - Supports `&&`, `||` without escaping.
  - Allows regex matching with `=~`.

---



## Checking File Conditions

- `-e file`: File exists.
- `-f file`: File is a regular file.
- `-d file`: Directory exists.
- `-s file`: File is not empty.
- `-r file`: File is readable.
- `-w file`: File is writable.
- `-x file`: File is executable.

---

## Practical Example: User Input Validation

```bash
read -p "Enter a number between 1 and 10: " num
if [[ $num -ge 1 && $num -le 10 ]]
then
    echo "Valid input."
else
    echo "Invalid input."
fi
```

- **Explanation**:
  - Prompts user for input.
  - Validates if the number is within the specified range.

---



## Nested If Statements

- Place an if statement inside another for complex conditions.

```bash
if [ $user == "admin" ]
then
    if [ $password == "secret" ]
    then
        echo "Access granted."
    else
        echo "Incorrect password."
    fi
else
    echo "Unknown user."
fi
```



---

## Case Statement Alternative

- For multiple conditions, `case` can be more efficient.

```bash
case $option in
    start)
        echo "Starting service."
        ;;
    stop)
        echo "Stopping service."
        ;;
    restart)
        echo "Restarting service."
        ;;
    *)
        echo "Invalid option."
        ;;
esac
```



````python
%%bash
option=stop
case $option in
    start)
        echo "Starting service."
        ;;
    stop)
        echo "Stopping service."
        ;;
    restart)
        echo "Restarting service."
        ;;
    *)
        echo "Invalid option."
        ;;
esac
````

---

## Tips and Best Practices

- **Quote Variables**: Use `"$variable"` to prevent word splitting.
- **Use Double Brackets**: Prefer `[[ ]]` for complex conditions.
- **Indent Code**: Improves readability.
- **Test Conditions**: Use `echo` or `set -x` for debugging.

---

## Common Pitfalls

- **Incorrect Spacing**: `[ $a -gt $b ]` requires spaces.
- **Variable Expansion**: Uninitialized variables may cause errors.
- **Operator Confusion**: Remember `-eq` is for numbers, `=` is for strings.

- Can handle simple or complex conditions.
