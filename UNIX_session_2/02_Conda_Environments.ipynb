## What is Conda?

Conda is both:

- a package manager, which installs software and libraries; and
- an environment manager, which keeps project dependencies separate.

An environment gives us a named, repeatable collection of tools. For this module the environment is called `lectures`.


## Build an environment by hand

Before using a recipe file, let us create a small environment interactively. This makes each step visible.

In a terminal, run:

```bash
conda create --name unix-demo python=3.12
```

Read the package plan, then type `y` when Conda asks whether to proceed. What does Conda want to install, and where will it install it?


### Activate, add a package, and inspect

```bash
conda activate unix-demo
python --version
conda install pandas
conda list
```

Pause after each command:

1. How does the prompt show that `unix-demo` is active?
2. Which Python version is being used?
3. What extra packages were installed along with pandas?


### Leave the practice environment

This environment was created for practice, so it is safe to remove after the demonstration:

```bash
conda deactivate
conda env remove --name unix-demo
```

Manual installation is useful for learning and experimentation, but it is easy to forget a package or install slightly different versions. A YAML file records the recipe so the same environment can be recreated.


## Create the course environment

From a terminal, move into your course checkout and run:

```bash
cd LIFE4138_2627
conda env create -f lectures.yml
conda activate lectures
```

If the environment already exists, update it instead:

```bash
conda env update -f lectures.yml --prune
conda activate lectures
```


### Check that activation worked

The prompt usually shows the active environment in parentheses:

```bash
conda env list
which python
python --version
```

The `*` in `conda env list` marks the active environment.


## Read the recipe

Open `lectures.yml` in the repository. It is the recipe used to build the environment.

Look for:

- the environment name;
- the channels Conda searches;
- packages installed by Conda; and
- packages installed through the nested `pip` list.

Before looking at the answer, identify two packages in the file that you have not used yet. What might each one be useful for?


### Inspect the installed packages

```bash
conda list
conda list | grep -E 'cowsay|pandas|matplotlib|seaborn|biopython'
```

The first command shows the complete environment. The second filters it to packages that may be new to you.


## A small package with a big personality
