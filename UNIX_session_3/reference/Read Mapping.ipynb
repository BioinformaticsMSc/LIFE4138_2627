# Read-mapping command reference

This notebook is deliberately a compact reference, not a third lesson. Run the guided practical from `../01_Read_Mapping_and_Coverage.ipynb`.


## Where commands run

The notebook is stored beside the FASTA in `reference/`, but the course notebooks and scripts run from the parent `Unix3_improved` directory. The commands below therefore begin with `cd ..` when copied from this location.


## File products

| Product | Purpose |
|---|---|
| FASTA | Reference sequences |
| FASTQ.gz | Compressed reads and quality strings |
| Sorted BAM | Coordinate-ordered alignments |
| BAI | Index for region access |
| Flagstat text | Alignment-count summary |
| Depth TSV | Reference, position, and depth |


## Complete command sequence

```bash
cd ..
mkdir -p work/reference_example
run_dir=work/reference_example

minimap2 -t 2 -ax map-ont reference/HVol_Complete.fasta barcode09/*.fastq.gz \
  | samtools sort -@ 1 -o "$run_dir/aligned.sorted.bam" -
samtools index "$run_dir/aligned.sorted.bam"
samtools quickcheck -v "$run_dir/aligned.sorted.bam"
samtools flagstat "$run_dir/aligned.sorted.bam"
samtools depth -aa "$run_dir/aligned.sorted.bam" > "$run_dir/depth.tsv"
```


## Interpretation reminders

- A successful command does not prove the correct sample and reference were used.
- `depth -aa` includes entirely uncovered reference sequences; `-a` may omit them.
- Secondary and supplementary alignments require an explicit reporting decision.
- Mapping-quality and base-quality thresholds can alter depth.
- Record tool versions, commands, input identities, and checksums.
