## 1. Why does indentation matter?

Python uses **indentation** to define blocks of code. Use **four spaces** per level. A colon (`:`) introduces a block after statements such as `if`, `for`, and `def`.

````python
age = 10
if age >= 18:
    print("You are an adult.")
    print("You can vote.")
print("This message always appears.")
````

## 2. Indentation changes behaviour

Both examples are valid Python, but they behave differently. Predict their outputs before running them.

````python
temperature = 15
if temperature > 20:
    print("It is warm.")
    print("You might want a cold drink.")
print("Have a nice day!")
````

````python
temperature = 15
if temperature > 20:
    print("It is warm.")
print("You might want a cold drink.")
print("Have a nice day!")
````

## 3. Conditional statements

`if`, `elif`, and `else` must align when part of the same conditional. Their bodies are indented.

````python
score = 75
if score >= 70:
    print("Grade: A")
    print("Excellent work!")
elif score >= 60:
    print("Grade: B")
    print("Good work!")
else:
    print("Keep practising!")
print("Grading complete.")
````

## 4. Indentation in loops

Every statement inside a `for` loop is repeated. Statements outside execute after the loop.

````python
for i in range(1, 4):
    print("Number:", i)
    print("Square:", i ** 2)
print("Loop finished!")
````

````python
for i in range(1, 4):
    print("Number:", i)
print("Square:", i ** 2)
print("Loop finished!")
````

## 5. Nested indentation

Nested blocks add another four spaces per level.

````python
for number in range(1, 6):
    print("Checking number:", number)
    if number % 2 == 0:
        print("    This number is even.")
    else:
        print("    This number is odd.")
print("Finished checking numbers.")
````

````python
for i in range(1, 4):
    print("Outer loop:", i)
    for j in range(1, 4):
        print("    Inner loop:", j)
    print("Inner loop finished.")
print("All loops finished.")
````

## 9. Indentation in functions

Function bodies are indented. A return to the outer level marks the end of the function definition.

````python
def greet(name):
    print("Hello,", name)
    print("Welcome to Python!")
print("Starting program...")
greet("Alice")
greet("Bob")
print("Program finished.")
````

## 10. Key takeaways

**Remember:** four spaces per level; align statements in the same block; indentation changes execution, not just appearance.
